  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
# 202432021_Farel Mahandika
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import seaborn as sns
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>
<!-- 202432021_Farel Mahandika -->

```python
# 202432021_Farel Mahandika
# 1. Muat dataset ulasan e-commerce
# Pastikan file 20191002-reviews.csv berada di folder yang sama dengan notebook.
df = pd.read_csv('20191002-reviews.csv')

# Filter kolom penting dan hapus data yang tidak lengkap
df = df[['reviewContent', 'rating']].dropna().copy()
df['reviewContent'] = df['reviewContent'].astype(str)
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df = df.dropna(subset=['rating'])

# Label sentimen: rating 4-5 = positif (1), rating 1-3 = negatif (0)
df['sentiment'] = (df['rating'] > 3).astype(int)

# Ambil sampel agar proses pelatihan lebih efisien
df_sampled = df.sample(n=min(10000, len(df)), random_state=42).reset_index(drop=True)

# 2. Pembersihan teks
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df_sampled['cleaned_text'] = df_sampled['reviewContent'].apply(clean_text)
df_sampled = df_sampled[df_sampled['cleaned_text'].str.strip() != ''].reset_index(drop=True)

# 3. Pisahkan data terlebih dahulu agar vocabulary tidak mempelajari data testing.
# Sebanyak 80% menjadi data pengembangan dan 20% menjadi data testing.
train_text, test_text, y_train, y_test = train_test_split(
    df_sampled['cleaned_text'].values,
    df_sampled['sentiment'].values,
    test_size=0.2,
    random_state=42,
    stratify=df_sampled['sentiment'].values
)

# Sisihkan 10% dari data pengembangan untuk validasi.
train_text, val_text, y_train, y_val = train_test_split(
    train_text, y_train,
    test_size=0.1,
    random_state=42,
    stratify=y_train
)

# 4. Bangun vocabulary HANYA dari teks training.
# Indeks 0 digunakan untuk padding dan indeks 1 untuk kata di luar vocabulary (UNK).
all_train_words = ' '.join(train_text).split()
vocab_counts = Counter(all_train_words)
vocab_sorted = sorted(vocab_counts, key=vocab_counts.get, reverse=True)
vocab_to_int = {word: idx + 2 for idx, word in enumerate(vocab_sorted)}

# 5. Tokenisasi, sequence encoding, dan padding.
seq_length = 30

def encode_and_pad(texts, vocab, max_len=30):
    features = np.zeros((len(texts), max_len), dtype=np.int64)
    for i, text in enumerate(texts):
        tokens = [vocab.get(word, 1) for word in text.split()]
        tokens = tokens[-max_len:]  # ambil maksimal max_len token terakhir
        if tokens:
            features[i, -len(tokens):] = tokens  # padding di sebelah kiri
    return features

X_train = encode_and_pad(train_text, vocab_to_int, seq_length)
X_val = encode_and_pad(val_text, vocab_to_int, seq_length)
X_test = encode_and_pad(test_text, vocab_to_int, seq_length)

# 6. Konversi ke tensor PyTorch
def make_loader(features, labels, batch_size=64, shuffle=False):
    x_tensor = torch.tensor(features, dtype=torch.long)
    y_tensor = torch.tensor(labels, dtype=torch.float32)
    dataset = TensorDataset(x_tensor, y_tensor)
    return DataLoader(dataset, batch_size=batch_size, shuffle=shuffle)

train_loader = make_loader(X_train, y_train, shuffle=True)
val_loader = make_loader(X_val, y_val)
test_loader = make_loader(X_test, y_test)

print('Preprocessing selesai.')
print(f'Total data setelah pembersihan: {len(df_sampled):,}')
print(f'Data training: {len(X_train):,} ({len(X_train)/len(df_sampled)*100:.1f}%)')
print(f'Data validasi: {len(X_val):,} ({len(X_val)/len(df_sampled)*100:.1f}%)')
print(f'Data testing : {len(X_test):,} ({len(X_test)/len(df_sampled)*100:.1f}%)')
print(f'Ukuran vocabulary (di luar PAD dan UNK): {len(vocab_to_int):,}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>
<!-- 202432021_Farel Mahandika -->

```python
# 202432021_Farel Mahandika
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        self.rnn = nn.RNN(embedding_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, output_dim)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        embeds = self.embedding(x)
        _, hidden = self.rnn(embeds)
        output = self.fc(hidden.squeeze(0))
        # squeeze(1) menjaga dimensi batch tetap ada, termasuk saat batch berisi 1 data.
        return self.sigmoid(output).squeeze(1)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>
<!-- 202432021_Farel Mahandika -->

```python
# 202432021_Farel Mahandika
# 1. Inisiasi hyperparameter
vocab_size = len(vocab_to_int) + 2  # indeks 0=PAD dan 1=UNK
embedding_dim = 64
hidden_dim = 32
output_dim = 1
epochs = 20
learning_rate = 0.001

# 2. Inisiasi model, loss function, dan optimizer
model = SequenceModel(vocab_size, embedding_dim, hidden_dim, output_dim)
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Fungsi untuk menghitung loss pada data validasi
def calculate_loss(data_loader):
    model.eval()
    total_loss = 0.0
    with torch.no_grad():
        for inputs, labels_batch in data_loader:
            outputs = model(inputs)
            loss = criterion(outputs, labels_batch)
            total_loss += loss.item() * inputs.size(0)
    return total_loss / len(data_loader.dataset)

train_losses = []
val_losses = []

# 4. Training loop dengan pemantauan validation loss
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for inputs, labels_batch in train_loader:
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels_batch)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * inputs.size(0)

    epoch_train_loss = running_loss / len(train_loader.dataset)
    epoch_val_loss = calculate_loss(val_loader)
    train_losses.append(epoch_train_loss)
    val_losses.append(epoch_val_loss)

    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f'Epoch [{epoch+1:02d}/{epochs}] | Training Loss: {epoch_train_loss:.4f} | Validation Loss: {epoch_val_loss:.4f}')

print('Pelatihan model selesai.')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. EVALUASI MODEL, VISUALISASI LOSS & PERBANDINGAN PREDIKSI</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mengevaluasi model menggunakan accuracy, precision, recall, F1-score, confusion matrix, serta visualisasi loss dan hasil prediksi.</p>
</div>
<!-- 202432021_Farel Mahandika -->

```python
# 202432021_Farel Mahandika
# 1. Grafik training loss dan validation loss
plt.figure(figsize=(10, 4))
plt.plot(range(1, epochs + 1), train_losses, linewidth=2, label='Training Loss')
plt.plot(range(1, epochs + 1), val_losses, linewidth=2, label='Validation Loss')
plt.title('Training Loss dan Validation Loss Model RNN')
plt.xlabel('Epoch')
plt.ylabel('Binary Cross-Entropy Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

# 2. Prediksi pada data testing (data yang tidak digunakan saat training/validasi)
model.eval()
probabilities = []
actuals = []
with torch.no_grad():
    for inputs, labels_batch in test_loader:
        outputs = model(inputs)
        probabilities.extend(outputs.cpu().numpy().tolist())
        actuals.extend(labels_batch.cpu().numpy().astype(int).tolist())

predictions = (np.array(probabilities) >= 0.5).astype(int)
actuals = np.array(actuals).astype(int)

# 3. Evaluasi klasifikasi
accuracy = accuracy_score(actuals, predictions)
precision = precision_score(actuals, predictions, zero_division=0)
recall = recall_score(actuals, predictions, zero_division=0)
f1 = f1_score(actuals, predictions, zero_division=0)

print('HASIL EVALUASI MODEL RNN PADA DATA TESTING')
print(f'Accuracy  : {accuracy:.4f} ({accuracy*100:.2f}%)')
print(f'Precision : {precision:.4f} ({precision*100:.2f}%)')
print(f'Recall    : {recall:.4f} ({recall*100:.2f}%)')
print(f'F1-score  : {f1:.4f} ({f1*100:.2f}%)')
print('\nClassification Report:')
print(classification_report(actuals, predictions, labels=[0, 1], target_names=['Negatif', 'Positif'], zero_division=0))

# 4. Confusion matrix
cm = confusion_matrix(actuals, predictions, labels=[0, 1])
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Negatif', 'Positif'],
            yticklabels=['Negatif', 'Positif'])
plt.title('Confusion Matrix Model RNN')
plt.xlabel('Prediksi Model')
plt.ylabel('Label Aktual')
plt.tight_layout()
plt.show()

# 5. Perbandingan 50 sampel pertama pada data testing
sample_size = min(50, len(actuals))
plt.figure(figsize=(12, 5))
plt.plot(actuals[:sample_size], label='Aktual', marker='o', linewidth=2)
plt.plot(predictions[:sample_size], label='Prediksi RNN', linestyle='--', marker='x', linewidth=2)
plt.title(f'Perbandingan Sentimen Aktual vs Prediksi RNN ({sample_size} Sampel Testing)')
plt.xlabel('Sampel Ulasan')
plt.ylabel('Kelas Sentimen (1: Positif, 0: Negatif)')
plt.yticks([0, 1], ['Negatif', 'Positif'])
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
